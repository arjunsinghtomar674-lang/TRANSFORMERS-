```python
import tensorflow as tf
```

```python
tf.random.set_seed(42)
```

```python
X = tf.random.normal((2,4,8))
```

```python
num_heads = 2
key_dim = 4
```

```python
multihead_attention = tf.keras.layers.MultiHeadAttention(
    num_heads = num_heads,
    key_dim = key_dim
)
```

```python
output,attention_scores = multihead_attention(
    query = X,
    value = X,
    key = X,
    return_attention_scores=True
)
```

```python
print("output shape",output.shape)
```

```python
print("attention shape",attention_scores.shape)
```

```python
final_input = X + output
```

```python
print(final_input.shape)
```

```python
add_norm1 = tf.keras.layers.LayerNormalization()
attention_output = add_norm1(final_input)
```

```python
print(attention_output.shape)
```

```python
ffnn = tf.keras.Sequential([
    tf.keras.layers.Dense(16,activation = "relu"),
    tf.keras.layers.Dense(8)
])
```

```python
ffnn_output = ffnn(attention_output)
```

```python
print(ffnn_output.shape)
```

```python
add_norm2 = tf.keras.layers.LayerNormalization()
encoder_output = add_norm2(ffnn_output + attention_output)
```

```python
print(encoder_output.shape)
```
