```python
import tensorflow as tf
```

```python
tf.random.set_seed(42)
```

```python
X = tf.random.normal((1,4,8))
```

```python
num_heads = 2
key_dim = 4
```

```python
multihead_attention = tf.keras.layers.MultiHeadAttention(
    num_heads = num_heads,
    key_dim = key_dim
)
```

```python
output,attention_score = multihead_attention(
    query = X,
    value = X,
    key = X,
    return_attention_scores = True
)
```

```python
print("input shape",X.shape)
print("output shape",output.shape)
print("attention score shape",attention_score.shape)
```
