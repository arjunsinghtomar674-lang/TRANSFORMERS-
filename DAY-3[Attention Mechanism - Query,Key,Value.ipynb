# Day 3: Attention Mechanism — Query, Key, and Value

```python
import tensorflow as tf
```

```python
tf.random.set_seed(42)
```

```python
X = tf.constant([
    [1.0, 0.0, 1.0, 0.0],
    [0.0, 2.0, 0.0, 2.0],
    [1.0, 1.0, 1.0, 1.0]
])

print("Input shape:", X.shape)
```

```python
d_model = 4
d_k = 2
```

```python
W_q =  tf.keras.layers.Dense(d_k,use_bias=False)
W_k = tf.keras.layers.Dense(d_k,use_bias=False)
W_v = tf.keras.layers.Dense(d_k,use_bias=False)
```

```python
Q = W_q(X)
K = W_k(X)
V = W_v(X)
```

```python
print(Q.shape)
```

```python
scores = tf.matmul(Q,K,transpose_b = True)
print("Attention scores:")
print(scores.numpy())

print("Scores shape:", scores.shape)
```
