## 🎯 Learning Objectives
- Understand tokens, vocabulary, and token IDs.
- Convert text into numerical sequences using `TextVectorization`.
- Understand padding and padding masks.
- Create embeddings using `Embedding`.
- Understand the shape `(batch_size, sequence_length, embedding_dim)`.

```
import tensorflow as tf
from tensorflow.keras.layers import TextVectorization
```

```
sentences = [
    "I love machine learning",
    "I love deep learning",
    "Machine learning is interesting",
    "Deep learning uses neural networks"
]
```

```
tokenizer = TextVectorization(
    output_mode = "int"
)
tokenizer.adapt(sentences)
```

```
tokenizer = TextVectorization(
    output_mode="int",
    output_sequence_length=6
)

tokenizer.adapt(sentences)

token_ids = tokenizer(
    tf.constant([
        "I love AI",
        "I love deep learning"
    ])
)

print(token_ids.numpy())
print("Shape:", token_ids.shape)
```

```
embedding_dim = 8
```

```
embedding_layer = tf.keras.layers.Embedding(
    input_dim = len(vocabulary),
    output_dim = embedding_dim,
    mask_zero = True
)
```

```
embeddings = embedding_layer(token_ids)
```
